앞부분은 본문 순서에 맞춘 실습이다. 뒤에서는 번호 목록, 뷰와 복사본, 조건 결합 등을 실습한다.

## 같은 열을 골라도 결과의 모양은 달라진다

```python
import numpy as np

grid = np.array([[1, 2, 3], [4, 5, 6]])
column_values = grid[:, 1]
column_array = grid[:, 1:2]

print(column_values, column_values.shape)
print(column_array, column_array.shape)
```

## 위치 하나와 범위를 지정하는 방법이 다르다

```python
import numpy as np

a = np.array([1, 2, 3, 4, 5, 6])

print(a[0])
print(a[:1], a[:1].shape)
print(a[1:3])
```

## 3차원 배열에서는 고른 뒤 남는 축을 읽는다

```python
import numpy as np

t = np.arange(24).reshape(2, 3, 4)
selected = t[:, 0, :2]
kept_rows = t[:, 0:1, :2]

print(selected, selected.shape, sep="\n")
print(kept_rows, kept_rows.shape, sep="\n")
```

## 마스크의 참 하나가 점수 한 칸 또는 학생 한 행을 고른다

```python
import numpy as np

scores = np.array([[72, 88, 64, 90], [55, 67, 81, 70], [91, 74, 82, 96]])
cell_mask = scores >= 80
row_mask = scores[:, 0] >= 70

print(cell_mask)
print(scores[cell_mask], scores[cell_mask].shape)
print(row_mask)
print(scores[row_mask], scores[row_mask].shape)
```

### 직접 확인해 보기

```python
import numpy as np

scores = np.array([[72, 88, 64, 90], [55, 67, 81, 70], [91, 74, 82, 96]])
row_mask = scores[:, 0] >= 90
selected = scores[row_mask]
print(row_mask)
print(selected, selected.shape)
```

## 보충 실습

### 뒤에서 세거나 일정한 간격으로 고르기

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])
print(a[-1])
print(a[:3])
print(a[3:])
print(a[::2])
```

### 한 칸과 행·열 범위 고르기

```python
import numpy as np

c = np.array([[1, 2, 3], [4, 5, 6]])
print(c[1, 2])
print(c[0, :])
print(c[:, 1:3])
```

### 슬라이스의 뷰와 복사본 비교하기

```python
import numpy as np

c = np.array([[1, 2, 3], [4, 5, 6]])
column = c[:, 1]
separate = c[:, 1].copy()
column[0] = 20
separate[1] = 50
print(c)
print(separate)
```

### 표 한 장에서 값 하나까지 고르기

```python
import numpy as np

t = np.arange(24).reshape(2, 3, 4)
print(t.shape)
print(t[0])
print(t[0, 1])
print(t[0, 1, 2])
```

### 번호 목록으로 행과 열 고르기

```python
import numpy as np

table = np.arange(1, 13).reshape(3, 4)
print(table)
print(table[[2, 0], :])
print(table[1, [0, 2]])
```

### 좌표 두 개와 행·열의 모든 조합 비교하기

```python
import numpy as np

table = np.arange(1, 13).reshape(3, 4)
rows = [2, 0]
cols = [1, 3]
print(table[rows, cols])
print(table[np.ix_(rows, cols)])
```

### 복사본 수정과 원본 대입 비교하기

```python
import numpy as np

table = np.arange(1, 13).reshape(3, 4)
picked = table[1, [0, 1]]
picked[:] = [99, 100]
print(table[1])

table[1, [0, 1]] = [99, 100]
print(table[1])
```

### 마스크로 점수만 고르기

```python
import numpy as np

scores = np.array([55, 42, 77, 93, 35])
mask = scores > 50
print(mask)
print(scores[mask])
```

### 두 조건을 결합해 학생 고르기

```python
import numpy as np

scores = np.array([[72, 88, 64, 90], [55, 67, 81, 70], [91, 74, 82, 96]])
mask = (scores[:, 0] >= 70) & (scores[:, 1] < 80)
selected = scores[mask, 1:3]
print(mask)
print(selected, selected.shape)
```

### 3차원 슬라이스와 조건 바꾸기

```python
import numpy as np

t = np.arange(24).reshape(2, 3, 4)
chosen = t[:, 1, -2:]
print(chosen, chosen.shape, sep="\n")

scores = np.array([55, 42, 77, 93, 35])
print(scores[scores > 77])
print(scores[scores >= 77])
```
