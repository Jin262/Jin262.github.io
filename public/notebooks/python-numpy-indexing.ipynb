## 1차원 배열에서 값과 구간 고르기

```python
import numpy as np

a = np.array([1, 2, 3])
print(a[0])
print(a[0:2])
```

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])
print(a[-1])
print(a[:3])
print(a[3:])
print(a[::2])
```

## 행과 열로 값과 구간 고르기

### 한 칸과 열 범위 고르기

```python
import numpy as np

c = np.array([[1, 2, 3], [4, 5, 6]])
print(c[1, 2])
print(c[:, 1:3])
```

### 열 하나를 골라도 모양은 달라진다

```python
import numpy as np

c = np.array([[1, 2, 3], [4, 5, 6]])
print(c[0, :])
print(c[:, 1], c[:, 1].shape)
print(c[:, 1:2], c[:, 1:2].shape)
```

### 슬라이스의 뷰와 복사본 비교하기

```python
import numpy as np

c = np.array([[1, 2, 3], [4, 5, 6]])
column = c[:, 1]
separate = c[:, 1].copy()
column[0] = 20
separate[1] = 50
print(c)
print(separate)
```

## 3차원 배열에서 표·행·열 고르기

### 표 한 장에서 값 하나까지 고르기

```python
import numpy as np

t = np.arange(24).reshape(2, 3, 4)
print(t.shape)
print(t[0])
print(t[0, 1])
print(t[0, 1, 2])
```

### 같은 값을 고르면서 행 축 유지하기

```python
import numpy as np

t = np.arange(24).reshape(2, 3, 4)
selected = t[:, 0, :2]
kept_rows = t[:, 0:1, :2]
print(selected, selected.shape, sep="\n")
print(kept_rows, kept_rows.shape, sep="\n")
```

## 번호 목록으로 값 고르기

### 목록에 적은 순서대로 행 고르기

```python
import numpy as np

table = np.arange(1, 13).reshape(3, 4)
print(table)
print(table[[2, 0], :])
print(table[1, [0, 2]])
```

### 좌표 두 개와 행·열의 모든 조합 비교하기

```python
import numpy as np

table = np.arange(1, 13).reshape(3, 4)
rows = [2, 0]
cols = [1, 3]
print(table[rows, cols])
print(table[np.ix_(rows, cols)])
```

### 복사본 수정과 원본에 직접 대입하기

```python
import numpy as np

table = np.arange(1, 13).reshape(3, 4)
picked = table[1, [0, 1]]
picked[:] = [99, 100]
print(table[1])

table[1, [0, 1]] = [99, 100]
print(table[1])
```

## 불리언 마스크로 값 고르기

```python
import numpy as np

scores = np.array([55, 42, 77, 93, 35])
mask = scores > 50
print(mask)
print(scores[mask])
```

## 칸별 조건과 행별 조건 비교하기

```python
import numpy as np

scores = np.array([[72, 88, 64, 90], [55, 67, 81, 70], [91, 74, 82, 96]])
cell_mask = scores >= 80
row_mask = scores[:, 0] >= 70

print(cell_mask)
print(scores[cell_mask], scores[cell_mask].shape)
print(row_mask)
print(scores[row_mask], scores[row_mask].shape)
```

```python
import numpy as np

scores = np.array([[72, 88, 64, 90], [55, 67, 81, 70], [91, 74, 82, 96]])
mask = (scores[:, 0] >= 70) & (scores[:, 1] < 80)
selected = scores[mask, 1:3]
print(mask)
print(selected, selected.shape)
```

## 직접 확인해 보기

3차원 슬라이스와 조건 바꾸기

```python
import numpy as np

t = np.arange(24).reshape(2, 3, 4)
chosen = t[:, 1, -2:]
print(chosen, chosen.shape, sep="\n")

scores = np.array([55, 42, 77, 93, 35])
print(scores[scores > 77])
print(scores[scores >= 77])
```
