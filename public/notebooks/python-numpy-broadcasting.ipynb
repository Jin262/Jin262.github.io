## 배열 모양에 따른 덧셈 비교하기

```python
import numpy as np

a = np.array([1, 2, 3])
c = np.array([[1, 2, 3], [4, 5, 6]])

print(a + c)
try:
    a + c.reshape(3, 2)
except ValueError as error:
    print(f"ValueError: {error}".rstrip())
```

## 원소별 사칙연산

```python
import numpy as np

left = [1, 2, 3]
right = [4, 5, 6]
a = np.array(left)
b = np.array(right)

print(left + right)
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
```

```python
import numpy as np

values = np.array([7, -7])
print(values / 4)
print(values // 4)
print(values % 4)
```

## 행과 열에 브로드캐스팅하기

```python
import numpy as np

c = np.array([[1, 2, 3], [4, 5, 6]])
columns = np.array([10, 20, 30])
rows = np.array([100, 200])

print(c + columns)
try:
    print(c + rows)
except ValueError as error:
    print(f"ValueError: {error}".rstrip())
print(c + rows.reshape(2, 1))
```

## 모양 바꾸기와 축 바꾸기

```python
import numpy as np

numbers = np.arange(1, 13)
print(numbers.reshape(3, -1))
print(numbers.reshape(2, 6))
try:
    print(numbers.reshape(5, 3))
except ValueError as error:
    print(f"ValueError: {error}".rstrip())
```

```python
import numpy as np

c = np.array([[1, 2, 3], [4, 5, 6]])
print(c.reshape(3, 2))
print(c.T)

v = np.array([1, 2, 3])
print(v.T.shape, v.reshape(-1, 1).shape)
```

## 직접 확인해 보기

축이 맞는 덧셈과 3차원 전치

```python
import numpy as np

a = np.array([1, 2, 3])
c = np.array([[1, 2, 3], [4, 5, 6]])
print((a + c).shape)
print(a.reshape(3, 1) + c.reshape(3, 2))
try:
    a.reshape(3, 1) + c
except ValueError as error:
    print(f"ValueError: {error}".rstrip())

t = np.arange(24).reshape(2, 3, 4)
print((t + np.array([0, 10, 20, 30]))[0, 0])
swapped = t.transpose(0, 2, 1)
print(swapped.shape, t.T.shape)
print(t[1, 0, 2], swapped[1, 2, 0])
```
