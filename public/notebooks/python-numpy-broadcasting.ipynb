앞부분은 본문의 행·열별 덧셈 실습이다. 뒤에서는 다른 산술 연산, 3차원 배열, 모양 변경과 전치를 다룬다.

## 같은 표에 값 세 개는 더할 수 있고 두 개는 오류가 난다

```python
import numpy as np

grid = np.array([[1, 2, 3], [4, 5, 6]])
columns = np.array([10, 20, 30])
rows = np.array([100, 200])

print(grid + columns)

try:
    print(grid + rows)
except ValueError as error:
    print(f"ValueError: {error}".rstrip())
```

## 브로드캐스팅은 모양을 오른쪽 끝부터 비교한다

두 모양을 오른쪽 끝에 맞춘다. 축이 적은 배열의 앞에는 비교할 때 길이 1인 축이 있다고 본다. 대응하는 길이가 같거나 둘 중 하나가 1이면 맞는다.

## 행에 하나씩 더할 값은 열이 하나인 배열로 만든다

```python
import numpy as np

grid = np.array([[1, 2, 3], [4, 5, 6]])
rows = np.array([100, 200])
row_values = rows.reshape(2, 1)

print(row_values)
print(row_values.shape)
print(grid + row_values)
```

## 행별 값과 열별 값을 더해 새 표를 만든다

```python
import numpy as np

rows = np.array([100, 200]).reshape(2, 1)
columns = np.array([10, 20, 30])
result = rows + columns

print(result)
print(result.shape)
```

### 직접 확인해 보기

```python
import numpy as np

rows = np.array([100, 300]).reshape(2, 1)
columns = np.array([10, 20, 30])
result = rows + columns
print(result)
print(result.shape)
```

## 보충 실습

### 같은 세 값을 다른 모양의 표에 더하기

```python
import numpy as np

a = np.array([1, 2, 3])
c = np.array([[1, 2, 3], [4, 5, 6]])

print(a + c)
try:
    a + c.reshape(3, 2)
except ValueError as error:
    print(f"ValueError: {error}".rstrip())
```

### 원소별 사칙연산과 내림 나눗셈

```python
import numpy as np

left = [1, 2, 3]
right = [4, 5, 6]
a = np.array(left)
b = np.array(right)

print(left + right)
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
```

```python
import numpy as np

values = np.array([7, -7])
print(values / 4)
print(values // 4)
print(values % 4)
```

### 배열 모양 바꾸기와 축의 순서 바꾸기

```python
import numpy as np

numbers = np.arange(1, 13)
print(numbers.reshape(3, -1))
print(numbers.reshape(2, 6))
try:
    print(numbers.reshape(5, 3))
except ValueError as error:
    print(f"ValueError: {error}".rstrip())
```

```python
import numpy as np

c = np.array([[1, 2, 3], [4, 5, 6]])
print(c.reshape(3, 2))
print(c.T)

v = np.array([1, 2, 3])
print(v.T.shape, v.reshape(-1, 1).shape)
```

### 3차원 배열에서 덧셈과 전치 확인하기

```python
import numpy as np

a = np.array([1, 2, 3])
c = np.array([[1, 2, 3], [4, 5, 6]])
print((a + c).shape)
print(a.reshape(3, 1) + c.reshape(3, 2))
try:
    a.reshape(3, 1) + c
except ValueError as error:
    print(f"ValueError: {error}".rstrip())

t = np.arange(24).reshape(2, 3, 4)
print((t + np.array([0, 10, 20, 30]))[0, 0])
swapped = t.transpose(0, 2, 1)
print(swapped.shape, t.T.shape)
print(t[1, 0, 2], swapped[1, 2, 0])
```
