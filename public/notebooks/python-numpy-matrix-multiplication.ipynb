## 같은 배열에 `*`와 `@`를 쓰면 다른 결과가 나온다

```python
import numpy as np

m = np.array([[1, 2], [4, 5], [6, 7]])
n = np.array([[3, 2, 1], [6, 5, 4]])

try:
    print(m * n)
except ValueError as error:
    print(f"ValueError: {error}".rstrip())

print(m @ n)
```

## 결과의 한 칸은 왼쪽 한 행과 오른쪽 한 열로 계산한다

```python
import numpy as np

m = np.array([[1, 2], [4, 5], [6, 7]])
n = np.array([[3, 2, 1], [6, 5, 4]])

first_row = m[0, :]
first_column = n[:, 0]
products = first_row * first_column

print(products)
print(products.sum())
print((m @ n)[0, 0])
```

## 안쪽 길이로 곱할 수 있는지, 바깥쪽 길이로 결과 모양을 읽는다

```python
import numpy as np

m = np.array([[1, 2], [4, 5], [6, 7]])
n = np.array([[3, 2, 1], [6, 5, 4]])

print((m @ n).shape)

try:
    print(m @ m)
except ValueError as error:
    print(f"ValueError: {error}".rstrip())
```

## 곱하는 순서를 바꾸고 다른 칸을 계산해 본다

```python
import numpy as np

m = np.array([[1, 2], [4, 5], [6, 7]])
n = np.array([[3, 2, 1], [6, 5, 4]])
reversed_product = n @ m

print(reversed_product)
print(reversed_product.shape)
print((m @ n)[1, 2])
```

## 보충 실습

### 내적 계산과 기준 방향으로의 정사영

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a * b)
print(a @ b)
print(np.dot(a, b))
```

```python
import numpy as np

arrows = np.array([[3, 2], [0, 2], [-2, 2]])
x_unit = np.array([1, 0])
print(arrows @ x_unit)
print(arrows[0] @ np.array([2, 0]))
```

### 두 연산이 모두 가능한 배열에서 결과 비교하기

```python
import numpy as np

left = np.array([[1, 2], [3, 4]])
right = np.array([[5, 6], [7, 8]])
print(left * right)
print(left @ right)
```

### 1차원 배열과 한 열 배열을 각각 곱하기

```python
import numpy as np

rows = np.array([[1, 2, 3], [4, 5, 6]])
vector = np.array([3, 4, 5])
column = vector.reshape(3, 1)

flat_result = rows @ vector
column_result = rows @ column
print(flat_result, flat_result.shape)
print(column_result, column_result.shape, sep="\n")
```

### 모양이 다른 두 행렬곱 비교하기

```python
import numpy as np

a = np.array([[5, 4, 6], [3, 5, 4], [1, 2, 3]])
b = np.array([[1, 2], [3, 4], [5, 6]])
print(a @ b)

p = np.arange(1, 9).reshape(4, 2)
q = np.array([[1, 3, 5, 7], [2, 4, 6, 8]])
print(p @ q)
```

### 평면에서 좌표를 늘리거나 돌리기

```python
import numpy as np

points = np.array([[1, 1], [-1, 1], [-1, -1], [1, -1]])
scale = np.array([[2, 0], [0, 1]])
turn = np.array([[0, 1], [-1, 0]])

print((points @ scale).tolist())
print((points @ turn).tolist())
print(points[0] @ scale @ turn, points[0] @ turn @ scale)
```

### 3차원에서 z축을 기준으로 돌리기

```python
import numpy as np

points = np.array([[2, 1, 2], [0, 2, 1], [1, -1, 3]])
rotate_z = np.array([[0, 1, 0], [-1, 0, 0], [0, 0, 1]])
turned = points @ rotate_z

by_loop = []
for x, y, z in points:
    by_loop.append([-y, x, z])

print(turned)
print(turned.shape)
print(np.array_equal(turned, np.array(by_loop)))
```

```python
import numpy as np

point = np.array([2, 1, 2])
theta = np.radians(30)
c, s = np.cos(theta), np.sin(theta)
rotate_z = np.array([[c, s, 0], [-s, c, 0], [0, 0, 1]])
print(np.round(point @ rotate_z, 3))
```

### 직접 확인해 보기 · y축을 기준으로 돌리기

```python
import numpy as np

point = np.array([2, 1, 2])
rotate_y = np.array([[0, 0, -1], [0, 1, 0], [1, 0, 0]])
print(point @ rotate_y)
```
