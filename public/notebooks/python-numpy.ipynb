## 리스트로 1차원·2차원 배열 만들기

```python
import numpy as np

a = np.array([1, 2, 3])
c = np.array([[1, 2, 3], [4, 5, 6]])

print(a)
print(c)
```

## 모양, 차원 수, 원소 수, 자료형 확인하기

### 0차원·1차원·2차원 배열의 모양과 크기 비교하기

```python
import numpy as np

scalar = np.array(5)
line = np.array([1, 2, 3, 4, 5, 6])
c = np.array([[1, 2, 3], [4, 5, 6]])

print(scalar.ndim, scalar.shape, scalar.size)
print(line.ndim, line.shape, line.size)
print(c.ndim, c.shape, c.size)
```

### 1차원과 한 행·한 열로 만든 2차원 배열 비교하기

```python
import numpy as np

vector = np.array([1, 2, 3])
one_row = np.array([[1, 2, 3]])
one_col = np.array([[1], [2], [3]])
print(vector.shape, one_row.shape, one_col.shape)
print(vector[1], one_row[0, 1], one_col[1, 0])
```

### `dtype`으로 자료형 지정하기

```python
import numpy as np

values = np.array([1, 2, 3], dtype=float)
print(values)
print(values.ndim, values.shape, values.size, values.dtype)
```

### 자료형을 생략했을 때 결과 확인하기

```python
import numpy as np

mixed = np.array([1, 2.5, 3])
print(mixed)
print(mixed.dtype)
```

### 3차원 배열에서 표 한 장의 모양 확인하기

```python
import numpy as np

pages = np.zeros((2, 3, 4), dtype=int)
print(pages.ndim)
print(pages.shape)
print(pages[0].shape)
print(pages.size)
```

## 배열을 채우거나 일정한 간격으로 만들기

```python
import numpy as np

print(np.zeros((2, 3), dtype=int))
print(np.ones((2, 3), dtype=int))
print(np.eye(3, dtype=int))
```

```python
import numpy as np

print(np.arange(1, 7, 2))
print(np.linspace(0, 1, 5))
```

## 난수 배열 만들기

```python
import numpy as np

rng = np.random.default_rng(42)
uniform = rng.random((2, 3))
normal = rng.standard_normal((2, 3))
dice = rng.integers(1, 7, size=(2, 3))

print(np.round(uniform, 3))
print(np.round(normal, 3))
print(dice)
print(uniform.shape, normal.shape, dice.shape)
```

## 직접 확인해 보기

0과 1 사이의 점 개수, 0으로 채울 배열의 모양, 동전 결과를 담을 배열의 모양을 바꿔 본다.

```python
import numpy as np

print(np.linspace(0, 1, 3))
print(np.zeros((3, 2), dtype=int).shape)
rng = np.random.default_rng(42)
print(rng.integers(1, 3, size=(3, 2)))
```
