## 같은 여섯 값도 배치에 따라 모양이 달라진다

```python
import numpy as np

line = np.array([1, 2, 3, 4, 5, 6])
grid = np.array([[1, 2, 3], [4, 5, 6]])

print(line.shape)
print(grid.shape)
```

## 축의 개수와 길이, 전체 원소 수를 구분한다

```python
import numpy as np

scalar = np.array(5)
line = np.array([1, 2, 3, 4, 5, 6])
grid = np.array([[1, 2, 3], [4, 5, 6]])

print("scalar", scalar.ndim, scalar.shape, scalar.size)
print("line", line.ndim, line.shape, line.size)
print("grid", grid.ndim, grid.shape, grid.size)
```

## 같은 세 값도 한 줄·한 행·한 열의 모양이 다르다

```python
import numpy as np

line = np.array([1, 2, 3])
one_row = np.array([[1, 2, 3]])
one_col = np.array([[1], [2], [3]])

print(line.shape, line[1])
print(one_row.shape, one_row[0, 1])
print(one_col.shape, one_col[1, 0])
```

## 3차원 배열의 모양은 표·행·열 순서로 읽는다

```python
import numpy as np

pages = np.zeros((2, 3, 4), dtype=int)
first_page = pages[0]

print(pages.shape)
print(first_page)
print(first_page.shape)
```

## 모양을 정해 배열을 만들고 출력으로 확인한다

```python
import numpy as np

wide = np.zeros((2, 3), dtype=int)
tall = np.zeros((3, 2), dtype=int)

print(wide)
print(wide.shape, wide.size)
print(tall)
print(tall.shape, tall.size)
```

### 직접 확인해 보기

```python
import numpy as np

one_row = np.zeros((1, 6), dtype=int)
print(one_row)
print(one_row.shape, one_row.size)
```

## 보충 실습

### 자료형 지정과 추론

```python
import numpy as np

values = np.array([1, 2, 3], dtype=float)
print(values)
print(values.ndim, values.shape, values.size, values.dtype)
```

```python
import numpy as np

mixed = np.array([1, 2.5, 3])
print(mixed)
print(mixed.dtype)
```

### 배열을 채우거나 일정한 간격으로 만들기

```python
import numpy as np

print(np.zeros((2, 3), dtype=int))
print(np.ones((2, 3), dtype=int))
print(np.eye(3, dtype=int))
```

```python
import numpy as np

print(np.arange(1, 7, 2))
print(np.linspace(0, 1, 5))
```

### 난수 배열 만들기

```python
import numpy as np

rng = np.random.default_rng(42)
uniform = rng.random((2, 3))
normal = rng.standard_normal((2, 3))
dice = rng.integers(1, 7, size=(2, 3))

print(np.round(uniform, 3))
print(np.round(normal, 3))
print(dice)
print(uniform.shape, normal.shape, dice.shape)
```

### 지점 개수와 난수 배열의 모양 바꾸기

```python
import numpy as np

print(np.linspace(0, 1, 3))
rng = np.random.default_rng(42)
print(rng.integers(1, 3, size=(3, 2)))
```
