# Numpy5 · 축을 따라 집계하고 정렬하기

## 합계는 지정한 축의 값을 하나로 모은다

```python
import numpy as np

scores = np.array([[72, 88, 64, 90], [55, 67, 81, 70], [91, 74, 82, 96]])
by_subject = scores.sum(axis=0)
by_student = scores.sum(axis=1)

print(by_subject, by_subject.shape)
print(by_student, by_student.shape)
```

## 학생별 평균을 뺄 때는 집계한 축을 길이 1로 남긴다

```python
import numpy as np

scores = np.array([[72, 88, 64, 90], [55, 67, 81, 70], [91, 74, 82, 96]])
student_means = scores.mean(axis=1, keepdims=True)
differences = scores - student_means

print(scores.mean(axis=1).shape)
print(student_means)
print(student_means.shape)
print(differences[0])
```

## 정렬하면 같은 축 안에서 값의 순서가 바뀐다

```python
import numpy as np

scores = np.array([[72, 88, 64, 90], [55, 67, 81, 70], [91, 74, 82, 96]])
by_subject = np.sort(scores, axis=0)
by_student = np.sort(scores, axis=1)

print(by_subject)
print(by_student)
print(by_subject.shape, by_student.shape)
```

## 정렬할 위치를 구해 학생 기록을 함께 옮긴다

```python
import numpy as np

scores = np.array([[72, 88, 64, 90], [55, 67, 81, 70], [72, 74, 82, 96]])
names = np.array(["A", "B", "C"])
order = np.argsort(scores[:, 0], stable=True)

print(order)
print(names[order].tolist())
print(scores[order])
```

### 직접 확인해 보기 · 두 번째 과목으로 정렬하기

기준 열 번호를 1로 바꾼다. 이름 순서와 두 번째 과목의 점수를 확인한다.

```python
import numpy as np

scores = np.array([[72, 88, 64, 90], [55, 67, 81, 70], [72, 74, 82, 96]])
names = np.array(["A", "B", "C"])
order = np.argsort(scores[:, 1], stable=True)

print(order)
print(names[order].tolist())
print(scores[order])
print(scores[order, 1])
```

## 보충 실습

### 전체 합계와 전체 정렬

집계와 정렬에서 모든 칸을 한꺼번에 처리하면 출력이 어떻게 달라지는지 확인한다.

```python
import numpy as np

scores = np.array([[72, 88, 64, 90], [55, 67, 81, 70], [91, 74, 82, 96]])
print(scores.sum())
print(np.sort(scores, axis=None))
```

### 과목별 평균, 중앙값, 최솟값, 최댓값, 표준편차

```python
import numpy as np

scores = np.array([[72, 88, 64, 90], [55, 67, 81, 70], [91, 74, 82, 96]])
print(np.round(scores.mean(axis=0), 2))
print(np.median(scores, axis=0))
print(scores.min(axis=0))
print(scores.max(axis=0))
print(np.round(scores.std(axis=0), 2))
```

### 점수 하나를 바꾸고 평균과 중앙값 비교하기

```python
import numpy as np

scores = np.array([[72, 88, 64, 90], [55, 67, 81, 70], [91, 74, 82, 96]])
changed = scores.copy()
changed[2, 0] = 100
print(np.round(scores[:, 0].mean(), 2), np.median(scores[:, 0]))
print(np.round(changed[:, 0].mean(), 2), np.median(changed[:, 0]))
```

### 평균이 같은 두 점수 묶음의 표준편차 비교하기

기본값인 ddof=0과 분모를 하나 줄이는 ddof=1을 각각 적용한다.

```python
import numpy as np

practice = np.array([[70, 70, 70, 70], [60, 60, 80, 80]])
print(practice.mean(axis=1))
print(practice.std(axis=1))
print(np.round(practice.std(axis=1, ddof=1), 2))
```

### 정렬한 복사본과 원본을 바꾸는 정렬 비교하기

```python
import numpy as np

first = np.array([72, 55, 91])
print(np.sort(first))
print(first)
print(first.sort())
print(first)
```

### 행마다 다른 열 순서로 점수 꺼내기

각 행에서 얻은 열 번호를 같은 행에 적용한다.

```python
import numpy as np

scores = np.array([[72, 88, 64, 90], [55, 67, 81, 70], [91, 74, 82, 96]])
order_by_subject = np.argsort(scores, axis=1)
print(order_by_subject[0])
print(np.take_along_axis(scores, order_by_subject, axis=1))
```
